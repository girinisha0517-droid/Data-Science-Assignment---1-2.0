## Q3. Explain the following operators with examples

### (i) `/` — Division
The `/` operator performs division and normally returns a floating-point result.

Example: `10 / 3` gives `3.3333333333333335`.

### (ii) `%` — Modulus
The `%` operator returns the remainder after division.

Example: `10 % 3` gives `1`.

### (iii) `//` — Floor Division
The `//` operator performs division and returns the floor value of the result.

Example: `10 // 3` gives `3`.

### (iv) `**` — Exponentiation
The `**` operator raises one number to the power of another.

Example: `2 ** 3` gives `8`.

````python
a = 10
b = 3

print("Division (/):", a / b)
print("Modulus (%):", a % b)
print("Floor Division (//):", a // b)
print("Exponentiation (**):", 2 ** 3)
````

## Q5. Using a while loop, verify whether A is purely divisible by B and find how many times

````python
A = 20
B = 5

if B == 0:
    print("Division by zero is not allowed.")
else:
    count = 0
    remainder = A

    while remainder >= B:
        remainder -= B
        count += 1

    if remainder == 0:
        print(A, "is purely divisible by", B)
        print("It can be divided", count, "times.")
    else:
        print(A, "is not purely divisible by", B)
        print("Whole-number division gives", count, "times with remainder", remainder)
````

### Explanation
For `A = 20` and `B = 5`, repeated subtraction gives:

`20 → 15 → 10 → 5 → 0`

Therefore, 20 is purely divisible by 5, and it can be divided **4 times**.

## Q6. Create a list of 25 integers and print whether each element is divisible by 3

````python
numbers = list(range(1, 26))

for element in numbers:
    if element % 3 == 0:
        print(element, "is divisible by 3")
    else:
        print(element, "is not divisible by 3")
````

## Q7. Mutable and Immutable Data Types

### Mutable Data Types
A mutable data type can be changed after it has been created. For example, a list is mutable.

```python
my_list = [1, 2, 3]
my_list[0] = 100
```

After the change, `my_list` becomes `[100, 2, 3]`.

### Immutable Data Types
An immutable data type cannot be changed after it has been created. For example, a tuple is immutable.

Trying to change an element of a tuple causes an error.

### Examples
- **Mutable:** list, dictionary, set
- **Immutable:** string, tuple, integer, float

The following code demonstrates the property.

````python
# Mutable example
my_list = [1, 2, 3]
print("Before modification:", my_list)

my_list[0] = 100
print("After modification:", my_list)

# Immutable example
my_tuple = (1, 2, 3)
print("Tuple:", my_tuple)

try:
    my_tuple[0] = 100
except TypeError as e:
    print("Tuple cannot be modified:", e)
````
